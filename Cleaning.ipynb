```python
import pandas as pd
```

```python
ikea = pd.read_csv('ikea.csv', sep=',')
ikea
```

```python
ikea['InvoiceDate'] = pd.to_datetime(ikea['InvoiceDate'], dayfirst=True)
```

```python
ikea['invoice_dt'] = ikea['InvoiceDate'].dt.date
```

```python
ikea['time'] = ikea['InvoiceDate'].dt.time
```

```python
ikea
```

```python
ikea = ikea.drop(columns=['InvoiceDate'])
```

```python
ikea
```

```python
ikea.columns = ['invoiceno', 'stockcode', 'desc', 'quantity', 'unitprice', 'custid', 'country', 'date', 'time']
```

```python
ikea['custid'] = ikea['custid'].fillna(0)
```

```python
ikea = ikea[['invoiceno', 'date', 'time', 'stockcode', 'desc', 'quantity', 'unitprice', 'custid', 'country']]
```

```python
ikea
```

```python
ikea.to_csv('ikea_updated.csv')
```
